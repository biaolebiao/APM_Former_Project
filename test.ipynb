```python
# 测试代码示例
import torch
from models.apm_former import APM_Former_ImageOnly

```

```python
# 创建模型
model = APM_Former_ImageOnly(pretrained_swin_path=None)
model.eval()

# 模拟3D MRI输入 (B=2, C=1, D=96, H=96, W=96)
mri = torch.randn(2, 1, 96, 96, 96)

# 前向传播
with torch.no_grad():
    logits, aligned_feat, attn = model(mri)
    print(f"Logits shape: {logits.shape}")  # 应输出 (2, 2)
    print(f"Aligned features shape: {aligned_feat.shape}")  # 应输出 (2, 288, 24,24,24)
    print(f"Attention shape: {attn.shape}")  # 应输出 (2, 1, 24,24,24)
```
