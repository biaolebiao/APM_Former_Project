```python
import pandas as pd
import os
# 获取当前脚本所在目录
script_dir = os.getcwd()

# ==========================================
# 1. 读取数据
# ==========================================
print("正在读取 CSV 文件...")
df = pd.read_csv(os.path.join(script_dir, 'final_clinical_dataset.csv'))
sMCI = df[df['Label'] == 0].copy()
# print(sMCI)
sMCI['prefix_numeric'] = sMCI['SubjectID'].str.extract(r'(\d+)_S_').astype(int)
sMCI['suffix_numeric'] = sMCI['SubjectID'].str.extract(r'_S_(\d+)').astype(int)

sMCI_sorted = sMCI.sort_values(by=['prefix_numeric','suffix_numeric'])
sMCI_sorted.drop(columns=['prefix_numeric','suffix_numeric'], inplace=True)
sMCI_sorted = sMCI_sorted.reset_index(drop=True)
# print(sMCI_sorted)
sMCI_sorted.to_csv(os.path.join(script_dir, 'sMCI_sorted.csv'), index=False)
print("MCI_sorted.csv 文件已保存！")

pMCI = df[df['Label'] == 1].copy()
pMCI['prefix_numeric'] = pMCI['SubjectID'].str.extract(r'(\d+)_S_').astype(int)
pMCI['suffix_numeric'] = pMCI['SubjectID'].str.extract(r'_S_(\d+)').astype(int)

pMCI_sorted = pMCI.sort_values(by=['prefix_numeric','suffix_numeric'])
pMCI_sorted.drop(columns=['prefix_numeric','suffix_numeric'], inplace=True)
pMCI_sorted = pMCI_sorted.reset_index(drop=True)
# print(pMCI_sorted)
pMCI_sorted.to_csv(os.path.join(script_dir, 'pMCI_sorted.csv'), index=False)
print("pMCI_sorted.csv 文件已保存！")
```

```python
sMCI_subjects = sMCI_sorted['SubjectID'].to_list()
sMCI_subjects_str = ','.join(sMCI_subjects)
txt_file_path = os.path.join(script_dir, 'sMCI_subjects.txt')
with open(txt_file_path, 'w') as f:
    f.write(sMCI_subjects_str)
print("SubjectID已成功写入TXT文件。")

pMCI_subjects = pMCI_sorted['SubjectID'].to_list()
pMCI_subjects_str = ','.join(pMCI_subjects)
txt_file_path = os.path.join(script_dir, 'pMCI_subjects.txt')
with open(txt_file_path, 'w') as f:
    f.write(pMCI_subjects_str)
print("SubjectID已成功写入TXT文件。")



```
